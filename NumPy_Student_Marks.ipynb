# NumPy Student Marks

**Task:** create a NumPy array of student marks and calculate the mean, median, maximum, minimum and standard deviation.

**Objective:** learn how NumPy performs numerical analysis efficiently.

**Tools:** Python, NumPy, Jupyter Notebook

## 1. Import NumPy

```python
import numpy as np
```

## 2. Create the marks array

Marks (out of 100) for 10 students, created manually.

```python
marks = np.array([78, 85, 62, 91, 55, 88, 73, 95, 67, 80])
print("Student marks:", marks)
print("Number of students:", marks.size)
```

## 3. Calculate the statistics

```python
mean_marks = np.mean(marks)
median_marks = np.median(marks)
max_marks = np.max(marks)
min_marks = np.min(marks)
std_marks = np.std(marks)

print(f"Mean (average):     {mean_marks:.2f}")
print(f"Median (middle):    {median_marks:.2f}")
print(f"Maximum:            {max_marks}")
print(f"Minimum:            {min_marks}")
print(f"Standard deviation: {std_marks:.2f}")
```

## 4. Which students scored highest and lowest

```python
print("Top scorer: student", np.argmax(marks) + 1, "with", max_marks)
print("Lowest scorer: student", np.argmin(marks) + 1, "with", min_marks)
print("Students above the mean:", np.sum(marks > mean_marks))
```

## 5. Explanation of the results

- **Mean (77.40):** the average mark across the class. It adds every mark and divides by the number of students.
- **Median (79.00):** the middle mark when the marks are sorted. It is close to the mean, so the marks are not heavily skewed.
- **Maximum (95) and minimum (55):** the class spans 40 marks from the lowest to the highest score.
- **Standard deviation (12.40):** on average, marks sit about 12 points away from the mean, which shows a moderate spread of performance.

NumPy computed all of these with single function calls on the whole array, with no loops.

## 6. Interview questions

**What is NumPy?**
NumPy is a Python library for numerical computing. Its core object is the `ndarray`, a fast, fixed-type, multi-dimensional array, along with functions for maths, statistics and linear algebra.

**Why is NumPy useful in machine learning?**
ML works on large arrays of numbers (features, weights, images). NumPy runs operations on whole arrays in optimised C code (vectorisation), so it is far faster than Python loops, and libraries such as pandas, scikit-learn and TensorFlow are built on it or work directly with its arrays.

**What is standard deviation?**
A measure of how spread out values are from the mean. A small standard deviation means values sit close to the average; a large one means they are widely spread. `np.std()` calculates the population standard deviation by default.
